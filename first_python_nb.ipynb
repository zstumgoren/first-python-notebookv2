## Chapter 7 (Group By) - Grouping and Aggregating Data



```python
accident_list.head()
```

Count of accidents by make and model:

```python
accident_list.groupby("latimes_make_and_model").size()
```

Now tally the fatalities by make and model:

```python
accident_list.groupby("latimes_make_and_model")['total_fatalities'].sum()
```

```python
accident_counts = accident_list.groupby("latimes_make_and_model").size().rename("accidents").reset_index()
```

## Chapter 8 (Merging)

```python
# Grab the flight hours data from NTSB
survey = pd.read_csv("https://raw.githubusercontent.com/palewire/first-python-notebook/main/docs/src/_static/faa-survey.csv")
```

```python
merged_list
```

```python
accident_counts
```

Fix the `latimes_make_and_model` column in the survey data to match the accident counts data:

```python
# Don't forget upper() is buried in the "str" accessor/attribute on a pandas Series. You can't just call upper() on the Series itself. You have to use the str accessor first, like this:
survey["latimes_make_and_model"] = survey["latimes_make_and_model"].str.upper()
```

```python
survey
```

Now merge the two datasets together on the `latimes_make_and_model` column (which is now in the same format in both datasets):

```python
merged_list = pd.merge(accident_counts, survey, on="latimes_make_and_model")
```

```python
merged_list
```

```python
survey_row_count = len(survey)
accidents_row_count = len(accident_counts)
if survey_row_count != accidents_row_count:
    print(f"Warning: The number of rows in the survey data ({survey_row_count}) does not match the number of rows in the accident counts data ({accidents_row_count}).")
else:
    print("The number of rows in both datasets match. You're golden!")
```
